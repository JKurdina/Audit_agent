# Ретривер

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
import numpy as np
from typing import List, Dict
import mammoth
import pdfplumber
import os
import faiss
import pickle

class DocumentRetriever:
    def __init__(self, embedder, chunk_size=800, chunk_overlap=100):
        self.embedder = embedder
        self.chunks = []
        self.chunk_embeddings = None
        self.index_path = "index.1"
        self.my_chunks = "chunks.1"
        self.index = None
        self.dimension = None
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap

        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            separators=["\n\n", "\n", ". ", ", ", " ", ""],
            length_function=len,
            is_separator_regex=False
        )

        self._load_existing_db()
    
    def split_text(self, text: str) -> List[str]:
        """Разбивает текст на чанки"""
        return self.text_splitter.split_text(text)
    
    def add_document(self, text: str, metadata: Dict=None):
        """
        Добавляет новый документ в БД.
        Эмбеддинги создаются только для нового документа.
        """
        print("Разбиваем текст на чанки")
        new_chunks = self.split_text(text)

        if not new_chunks:
            print("Нет чанков для добавления")
            return
        
        print(f"Создание эмбеддингов для {len(new_chunks)} новых чанков...")
        new_embeddings = self.embedder.encode(
            new_chunks,
            batch_size=32,
            normalize_embeddings=True,
            show_progress_bar=True
        )

        if self.index is None:
            self.dimension = new_embeddings.shape[1]
            self.index = faiss.IndexFlatIP(self.dimension)
            print(f"Создан новый FAISS индекс размерности {self.dimension}")
        
        self.index.add(new_embeddings.astype('float32'))

        for chunk in new_chunks:
            self.chunks.append({
                'text': chunk,
                'metadata': metadata or {}
            })
        
        print(f"Добавлено {len(new_chunks)} чанков")
        print(f"Всего в БД: {len(self.chunks)} чанков, {self.index.ntotal} векторов")

        self._save_db()
    
    def _update_embeddings(self):
        """Обновление эмбеддингов всех чанков"""
        if not self.chunks:
            print("Нет чанков для создания эмбеддингов")
            return
        
        texts = [c["text"] for c in self.chunks]
        print(f"Создание эмбеддингов для {len(texts)} чанков...")

        self.chunk_embeddings = self.embedder.encode(
            texts,
            batch_size=16,
            normalize_embeddings=True,
            show_progress_bar=True
        )

        print(f"Эмбеддинги созданы: {self.chunk_embeddings.shape}")
    
    def search(self, query: str, top_k: int=3) -> List[Dict]:
        """Ищет релевантные чанки по запросу"""
        if self.index is None or self.index.ntotal == 0:
            print("БД пуста. Добавьте документы через add_document()")
            return []
        
        query_embedding = self.embedder.encode(
            [query],
            normalize_embeddigns=True
        )

        scores, indices = self.index.search(query_embedding.astype('float32'), top_k)

        results = []
        for score, idx in zip(scores[0], indices[0]):
            if idx != -1 and idx < len(self.chunks):
                results.append({
                    'text': self.chunks[idx]['text'],
                    'metadata': self.chunks[idx]['metadata'],
                    'similarity': float(score)
                })
        return results
    
    def _load_existing_db(self):
        """Загружает сохраненную БД из файлов"""
        faiss_file = f"{self.index_path}.faiss"
        chunks_file = f"{self.my_chunks}.pkl"

        if not os.path.exists(faiss_file):
            print(f"Файл {faiss_file} не найден. Будет создана новая БД.")
            return False
        if not os.path.exists(chunks_file):
            print(f"Файл {chunks_file} не найден. Будет создаана новая БД.")
            return False
        
        try:
            print(f"FAISS index file: {faiss_file}")
            self.index = faiss.read_index(faiss_file)
            self.dimension = self.index.d

            with open(chunks_file, 'rb') as f:
                data = pickle.load(f)
                self.chunks = data['chunks']

                if 'chunk_size' in data:
                    self.chunk_size = data['chunk_size']
                if 'chunk_overlap' in data:
                    self.chunk_overlap = data['chunk_overlap']
            
            print(f"БД загружена!")
            print(f"Векторов в FAISS: {self.index.ntotal}")
            print(f"Чанков в хранилище: {len(self.chunks)}")
            print(f"Размерность векторов: {self.dimension}")
            return True
        except Exception as e:
            print(f"Ошибка загрузки БД: {e}")
            return False
        
    def _save_db(self):
        """Сохраняет текущее состояние БД"""
        if self.index is None or self.index.ntotal == 0:
            print("Нет данных для сохранения")
            return
        
        faiss.write_index(self.index, f"{self.index_path}.faiss")

        with open(f"{self.my_chunks}.pkl", 'wb') as f:
            pickle.dump({
                'chunks': self.chunks,
                'chunk_size': self.chunk_size,
                'chunk_overlap': self.chunk_overlap
            }, f)
        
        print(f"БД сохранена: {self.index.ntotal} векторов, {len(self.chunks)} чанков")
    
    def get_stats(self) -> Dict:
        """Возвращает статистику БД"""
        return {
            'total_chunks': len(self.chunks),
            'total_vectors': self.index.ntotal if self.index else 0,
            'dimension': self.dimension,
            'chunk_size': self.chunk_size,
            'chunk_overlap': self.chunk_overlap,
            'index_path': self.index_path
        }
```

# Multi-query + reranking

```python
def expand_query(question: str) -> List[str]:
    return [
        question
    ]
def retrieve_with_rerank(question, retriever, reranker, top_k=8):
    queries = expand_query(question)
    all_results = []
    for q in queries:
        all_results.extend(retriever.search(q, top_k=8))
        unique = {r["text"]: r for r in all_results}.values()
        texts = [r["text"] for r in unique]

        pairs = [[question, t] for t in texts]
        scores = reranker.predict(pairs)
        ranked = sorted(
            zip(unique, scores),
            key=lambda x: x[1],
            reverse=True
        )
    return [r[0] for r in ranked[:top_k]]
```

# Текст промта, генерация ответа, вопрос агенту

```python
def build_promt(context, question):
    return f"""
    Ты - AI-агент для генерации аудиторских гипотез. Твоя задача - на основе предоставленных документов и данных предложить 3-5 обоснованных гипотез, которые могут быть использованы для дальнейшего аудиторского анализа.

    **Каждая гипотеза должна содержать:**
    - краткое описание сути гипотезы;
    - обоснование (почему она может быть актуальна, на какие данные или риски указывает);
    - возможные последствия для компании/проекта;
    - рекомендуемые шаги для проверки гипотезы.

    **Формат ответа:**
    1. Гипотеза 1: [описание]
        - Обоснование: [аргументы]
        - Последствия: [кратко]
        - Проверка: [шаги]
    2. Гипотеза 2:
        ...
    
    **Критерии**
    - гипотезы должны быть конкретными, релевантными и проверяемыми.
    - избегай общих формулировок, фокусируйся на реальных рисках и аномалиях.

    ***Контекст для анализа:*
    {context}.

    **Вопрос:**
    {question}.
"""
```

```python
def generate_answer(promt, model, tokenizer):
    messages = [
        {"role": "system", "content": "Ты-ассистент ведущего аудитора банковской группы."},
        {"role": "user", "content": promt}
    ]

    formatted_promt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_promt=True
    )

    inputs = tokenizer(formatted_promt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=2048,
            temperature=0.7,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id
        )
    
    generated_ids = outputs[0][inputs['input_ids'].shape[1]:]
    response = tokenizer.decode(generated_ids, skip_special_tokens=True)

    return response
```

```python
def ask_agent(question, retriever):
    print("Поиск + rerunk...")
    chunks = retrieve_with_rerank(question, retriever, reranker)
    context = "\n\n---\n\n".join([f"[ИСТОЧНИК {i+1}]\n{c["text"]}" for i, c in enumerate(chunks)])
    promt = build_promt(context, question)
    print("Анализ...")
    answer = generate_answer(promt, model, tokenizer)
    return answer, chunks
```

# Инициализация ретривера

```python
retriever = DocumentRetriever(
    embedder=embedder,
    chunk_size=800,
    chunk_overlap=100
)
```

```python
question = "Вычитка информации о финансовых транзакциях по бизнес-карте корпоративного клиента и их отображение по счету Клиента"
answer, used_chunks = ask_agent(question, retriever)

print("\n" + "="*50)
print("ОТВЕТ АГЕНТА:")
print("="*50)
print(answer)

print("\n" + "="*50)
print("ИСПОЛЬЗОВАННЫЕ ИСТОЧНИКИ:")
print("="*50)
for i, chunk in enumerate(used_chunks, 1):
    print(f"\nИсточник {i}:\n {chunk["metadata"]}")
    print(chunk["text"][:100] + "...")   
```
